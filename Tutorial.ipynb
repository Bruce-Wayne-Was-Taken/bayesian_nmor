```python
from matplotlib import pyplot as plt
import numpy as np
from random import random
```

```python
NTl = 1000
NPb = 0
tau = 3.053 * 60
h = 1.0
p = 1 - 2**(-h/tau)
tmax = 1000
```

```python
#brownian motion
L = 101

x = np.linspace(0, L-1, L)
y = np.linspace(0, L-1, L)

N = 1000

xcoordinate = [49]
ycoordinate = [49]

# need to chose a random direction from - up, down, right left
for i in range(N):
    moved = False
    while not moved:
        p = random()
        curr_x = xcoordinate[i]
        curr_y = ycoordinate[i]
        if p < 0.25:
            curr_x -= 1
            if curr_x < 0 or curr_x > L-1:
                continue
            moved = True
            xcoordinate.append(curr_x)
            ycoordinate.append(curr_y)
        elif p < 0.5:
            curr_y -= 1
            if curr_y < 0 or curr_y > L-1:
                continue
            moved = True
            xcoordinate.append(curr_x)
            ycoordinate.append(curr_y)
        elif p < 0.75:
            curr_x += 1
            if curr_x < 0 or curr_x > L-1:
                continue
            moved = True
            xcoordinate.append(curr_x)
            ycoordinate.append(curr_y)
        else:
            curr_y += 1
            if curr_y < 0 or curr_y > L-1:
                continue
            moved = True
            xcoordinate.append(curr_x)
            ycoordinate.append(curr_y)


plt.plot(xcoordinate, ycoordinate, marker = ".", markerfacecolor='blue',   # Sets the INSIDE marker colour (alias: mfc)
    markeredgecolor='green')
plt.scatter(xcoordinate[0],ycoordinate[0], s = 100, label = "start position", c = "blue")
plt.scatter(xcoordinate[-1],ycoordinate[-1], s = 100, label = "final position", c = "red")
# plt.xlim(0, 101)
# plt.ylim(0, 101)
plt.legend()
plt.show()
```
